# Sequence Labelling

## zero-shot

```python



def tokenize_with_punctuation(text):
    return re.findall(r"\w+|[^\w\s]", text)


def split_on_pattern(text):
    pattern = r"</?(?:claim|premise)>|[^\s<>]+"
    tokens = re.findall(pattern, text)
    return tokens
```

```python
def tokens_to_bio(tokens, size: int):
    clean_tokens = []
    bio_tags = []

    current_label = None
    is_begin = False

    for tok in tokens:
        # Opening tag
        if tok.startswith("<") and not tok.startswith("</"):
            current_label = tok[1:-1].lower()
            is_begin = True
            continue

        # Closing tag
        if tok.startswith("</"):
            current_label = None
            is_begin = False
            continue

        # Normal token
        clean_tokens.append(tok)

        if current_label is None:
            bio_tags.append("o")
        else:
            if is_begin:
                bio_tags.append(f"b-{current_label}")
                is_begin = False
            else:
                bio_tags.append(f"i-{current_label}")

    #return clean_tokens, bio_tags
    if len(bio_tags) < size:
        bio_tags += ["o"] * (size - len(bio_tags))
    return bio_tags[:size]
```

```python
def flatten(xss):
    return [x for xs in xss for x in xs]
```

# NER

## zero-shot

```python
import pandas as pd
from sklearn.metrics import f1_score
import os
```

```python



def tokenize_with_punctuation(text):
    return re.findall(r"\w+|[^\w\s]", text)


def split_on_pattern(text):
    pattern = r"</?\w+>|[^\s<>]+"
    tokens = re.findall(pattern, text)
    return tokens
```

```python
def tokens_to_bio(tokens, size: int = None):
    bio_tags = []

    for token in tokens:
        if token.startswith("<") and not token.startswith("</"):
            bio_tags.append(token[1:-1].lower())

    if size != None:
        bio_tags = bio_tags[:size]
        if len(bio_tags) < size:
            bio_tags += ["o"] * (size - len(bio_tags))

    return bio_tags
```

```python
def flatten(xss):
    return [x for xs in xss for x in xs]
```

```python
predictions = df.apply(lambda row: tokens_to_bio(split_on_pattern(row['response']), size=len(row['ner_tags'])),
                       axis=1).tolist()
predictions = flatten(predictions)
predictions[:5]
```

```python
round(f1_score(y_true=labels, y_pred=predictions, average="macro"), 3)
```
